# Spam Text Classification

```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
import re

def clean_text(text):
    text = text.lower()
    text = re.sub(r'http\S+', '', text)
    text = re.sub(r'\d+', '', text)
    text = re.sub(r'[^\w\s]', '', text)
    return text
```

```python
dataset = pd.read_csv('spam.tsv', sep = '\t')
```

```python
dataset.head()
```

```python
dataset.isnull().sum()
```

```python
dataset.dropna(inplace=True)
```

```python
ham = dataset[dataset['label']=='ham']
ham.head()
```

```python
spam = dataset[dataset['label']=='spam']
spam.head()
```

```python
ham.shape, spam.shape
```

```python
ham = ham.sample(spam.shape[0])
```

```python
ham.shape, spam.shape
```

```python
data = pd.concat([ham, spam], ignore_index=True)
```

```python
data.sample(5)
```

```python
data.shape
```

```python
x = data['message']
y = data['label']
```

```python
print(x)
```

```python
print(y)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)
```

```python
print(x_train)
```

```python
print(x_test)
```

```python
print(y_train)
```

```python
print(y_test)
```

```python
cv = CountVectorizer()
x_train_bow = cv.fit_transform(x_train.ravel())
x_test_bow = cv.transform(x_test.ravel())
```

```python
print(x_train_bow)
```

```python
print(x_test_bow)
```

```python
x_train_bow.shape
```

```python
tfidf = TfidfVectorizer()
x_train_tfidf = tfidf.fit_transform(x_train.ravel())
x_test_tfidf = tfidf.transform(x_test.ravel())
```

```python
print(x_train_tfidf)
```

```python
print(x_test_tfidf)
```

```python
x_train_tfidf.shape
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

lr = LogisticRegression()
lr.fit(x_train_bow, y_train)
y_pred_bow = lr.predict(x_test_bow)
```

```python
print("Accuracy (BoW):", accuracy_score(y_test, y_pred_bow))
print(classification_report(y_test, y_pred_bow))
```

```python
lr = LogisticRegression()
lr.fit(x_train_tfidf, y_train)
y_pred_tfidf = lr.predict(x_test_tfidf)
```

```python
print("Accuracy (TF-IDF):", accuracy_score(y_test, y_pred_tfidf))
print(classification_report(y_test, y_pred_tfidf))
```

```python
msg = ["Congratulations! You’ve won a free ticket!"]
msg_tfidf = tfidf.transform(msg)
print(lr.predict(msg_tfidf))
```
